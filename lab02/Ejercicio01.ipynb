```python
# a. True si es múltiplo de 3, False en caso contrario
es_multiplo_de_3 = lambda n: n % 3 == 0

# b. Devuelve el cubo de un número
cubo = lambda n: n ** 3

# c. Devuelve el producto de dos números
producto = lambda a, b: a * b

# d. Devuelve el mayor de dos números
mayor = lambda a, b: a if a >= b else b

# e. True si la palabra empieza con "A" o "a"
empieza_con_a = lambda palabra: palabra.lower().startswith("a")

# f. Convierte temperatura de Celsius a Fahrenheit
celsius_a_fahrenheit = lambda c: (c * 9 / 5) + 32

print("a. ¿9 es múltiplo de 3?:", es_multiplo_de_3(9))
print("b. Cubo de 4:", cubo(4))
print("c. Producto 6 * 7:", producto(6, 7))
print("d. Mayor entre 12 y 8:", mayor(12, 8))
print("e. ¿'Avión' empieza con A?:", empieza_con_a("Avión"))
print("f. 25 °C en Fahrenheit:", celsius_a_fahrenheit(25))

operaciones = [
    lambda x: x * 2,
    lambda x: x + 10,
    lambda x: x ** 2
]

numero = int(input("\nIngrese un número: "))

for operacion in operaciones:
    print(operacion(numero))
```
