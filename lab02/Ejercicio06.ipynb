```python
def movimientos_validos(piedras):
    """Retorna las cantidades válidas (1, 2 o 3) que se pueden retirar."""
    return [m for m in (1, 2, 3) if m <= piedras]


def minimax(piedras, es_turno_max, memo=None):
    """
    Evalúa el estado del juego desde la perspectiva de MAX (CPU).
    Retorna +1 si gana MAX, -1 si gana MIN.
    """
    if memo is None:
        memo = {}

    estado = (piedras, es_turno_max)
    if estado in memo:
        return memo[estado]

    if piedras == 0:
        return -1 if es_turno_max else 1

    if es_turno_max:
        mejor_valor = float("-inf")
        for quitar in movimientos_validos(piedras):
            valor = minimax(piedras - quitar, False, memo)
            if valor > mejor_valor:
                mejor_valor = valor
        memo[estado] = mejor_valor
        return mejor_valor
    else:
        peor_valor = float("inf")
        for quitar in movimientos_validos(piedras):
            valor = minimax(piedras - quitar, True, memo)
            if valor < peor_valor:
                peor_valor = valor
        memo[estado] = peor_valor
        return peor_valor


def elegir_jugada_cpu(piedras):
    """Selecciona el movimiento óptimo para MAX usando Minimax."""
    memo = {}
    mejor_jugada = 1
    mejor_valor = float("-inf")

    for quitar in movimientos_validos(piedras):
        valor = minimax(piedras - quitar, False, memo)
        if valor > mejor_valor:
            mejor_valor = valor
            mejor_jugada = quitar

    return mejor_jugada


def pedir_jugada_humano(piedras):
    """Solicita y valida la jugada del usuario (MIN)."""
    validos = movimientos_validos(piedras)
    while True:
        try:
            quitar = int(input(f"¿Cuántas piedras quitas? {validos}: "))
            if quitar in validos:
                return quitar
            print(f"Movimiento inválido. Elige una opción entre {validos}.")
        except ValueError:
            print("Por favor, ingresa un número entero válido.")


def jugar_nim(piedras_iniciales=10):
    piedras = piedras_iniciales
    turno_humano = True

    print(f"=== JUEGO NIM (Pila inicial: {piedras} piedras) ===")
    print("Reglas: En cada turno puedes quitar 1, 2 o 3 piedras.")
    print("Pierde el jugador que se quede sin movimientos.\n")

    while piedras > 0:
        print(f"Piedras restantes: {piedras}")

        if turno_humano:
            print(">> Turno de MIN (Humano)")
            quitar = pedir_jugada_humano(piedras)
            piedras -= quitar
            print(f"Quitaste {quitar} piedra(s).\n")
        else:
            print(">> Turno de MAX (Computadora)")
            quitar = elegir_jugada_cpu(piedras)
            piedras -= quitar
            print(f"La computadora quitó {quitar} piedra(s).\n")

        turno_humano = not turno_humano

    if turno_humano:
        print("No quedan piedras en tu turno. ¡Gana la Computadora (MAX)!")
    else:
        print("No quedan piedras en el turno de la computadora. ¡Ganaste (MIN)!")

if __name__ == "__main__":
    jugar_nim(10)
```
