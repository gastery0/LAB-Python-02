```python
numeros = [3, 10, 15, 22, 7, 12]
temperaturas_c = [0, 20, 37, 100]
palabras = ["sol", "programación", "python", "gato", "funcional", "luz"]
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# a. Lista con tuplas (n, n^2)
tuplas_cuadrado = [(n, n ** 2) for n in numeros]

# b. Cubos de los números mayores a 10
cubos_mayores_10 = [n ** 3 for n in numeros if n > 10]

# c. Solo los números pares
solo_pares = [n for n in numeros if n % 2 == 0]

# d. Formato ["valor: 10", "valor: 15", ...]
etiquetas_valor = [f"valor: {n}" for n in numeros]

# e. Conversión a grados Fahrenheit
temperaturas_f = [(c * 9 / 5) + 32 for c in temperaturas_c]

# f. Palabras con más de 5 letras
palabras_largas = [p for p in palabras if len(p) > 5]

# g. Solo las iniciales de las palabras
iniciales = [p[0] for p in palabras if p]

# h. Indicar "par" o "impar" para cada número
paridad = ["par" if n % 2 == 0 else "impar" for n in numeros]

# i. Aplanar una matriz (lista de listas)
matriz_aplanada = [elemento for fila in matriz for elemento in fila]

print("a.", tuplas_cuadrado)
print("b.", cubos_mayores_10)
print("c.", solo_pares)
print("d.", etiquetas_valor)
print("e.", temperaturas_f)
print("f.", palabras_largas)
print("g.", iniciales)
print("h.", paridad)
print("i.", matriz_aplanada)
```
