```python
from functools import reduce

palabras = ["casa", "elefante", "oso", "radar", "python", "ana", "árbol", "sol"]
datos_con_nulos = [10, None, "hola", None, 0, False, 42]
numeros = [5, 12, 25, 30, -15, 45, 8]
productos_dict = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

# a. Palabras con más de 4 letras
mas_de_4 = list(filter(lambda p: len(p) > 4, palabras))

# b. Eliminar los elementos nulos (None) preservando 0 y False
sin_nulos = list(filter(lambda x: x is not None, datos_con_nulos))

# c. Palabras que empiezan por una vocal
vocales = "aeiouáéíóú"
empiezan_vocal = list(filter(lambda p: p and p[0].lower() in vocales, palabras))

# d. Palabras que son palíndromos
palindromos = list(filter(lambda p: p.lower() == p.lower()[::-1], palabras))

# e. Números que terminan en 5
terminan_en_5 = list(filter(lambda n: abs(n) % 10 == 5, numeros))

# f. Productos con precio mayor a 100
productos_caros = list(filter(lambda prod: prod["precio"] > 100, productos_dict))

print("--- FILTER ---")
print("a.", mas_de_4)
print("b.", sin_nulos)
print("c.", empiezan_vocal)
print("d.", palindromos)
print("e.", terminan_en_5)
print("f.", productos_caros)

nums_reduce = [2, 3, 4, 5]
palabras_reduce = ["Programación", " ", "Funcional", " ", "en", " ", "Python"]

# a. Multiplicación de todos los números
multiplicacion_total = reduce(lambda acc, x: acc * x, nums_reduce)

# b. Concatenar una lista de palabras en una sola cadena
cadena_concatenada = reduce(lambda acc, p: acc + p, palabras_reduce)

# c. Hallar el mayor usando función auxiliar
def obtener_mayor(a, b):
    return a if a >= b else b

numero_mayor = reduce(obtener_mayor, nums_reduce)

print("\n--- REDUCE ---")
print("a.", multiplicacion_total)
print("b.", cadena_concatenada)
print("c.", numero_mayor)

lista_palabras = ["zorro", "manzana", "pera", "kiwi", "banano"]
lista_tuplas = [(3, "c"), (1, "a"), (4, "d"), (2, "b")]
productos_tuplas = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]
numeros_desordenados = [7, 2, 9, 4, 1, 6, 3, 8]
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]

# a. Ordenar palabras alfabéticamente
alfabetico = sorted(lista_palabras, key=lambda p: p.lower())

# b. Ordenar tuplas por el primer elemento
por_primer_elem = sorted(lista_tuplas, key=lambda t: t[0])

# c. Ordenar por precio (mayor a menor) y luego por nombre (A-Z)
productos_ordenados = sorted(productos_tuplas, key=lambda p: (-p[1], p[0]))

# d. Ordenar palabras según su longitud
por_longitud = sorted(lista_palabras, key=len)

# e. Primero los pares y luego los impares
pares_luego_impares = sorted(numeros_desordenados, key=lambda n: (n % 2, n))

# f. Contar frecuencia de letras en una palabra y ordenar el diccionario por frecuencia
palabra_objetivo = "programacion"
conteo = {}
for letra in palabra_objetivo.lower():
    conteo[letra] = conteo.get(letra, 0) + 1

conteo_ordenado = dict(sorted(conteo.items(), key=lambda item: item[1], reverse=True))

# g. Ordenar empleados por departamento y luego por edad
empleados_ordenados = sorted(empleados, key=lambda e: (e[1], e[2]))

print("\n--- SORTED ---")
print("a.", alfabetico)
print("b.", por_primer_elem)
print("c.", productos_ordenados)
print("d.", por_longitud)
print("e.", pares_luego_impares)
print("f.", conteo_ordenado)
print("g.", empleados_ordenados)
```
