```python
numeros = [2, 5, 10, 12, 15, 3, 20, 8]

# a. Filtrar números pares y luego elevarlos al cuadrado
pares_al_cuadrado = list(
    map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numeros))
)

# b. Elevar al cubo, filtrar los mayores que 100 y convertirlos a cadenas
cubos_mayores_100_str = list(
    map(str, filter(lambda x: x > 100, map(lambda x: x ** 3, numeros)))
)

# c. Filtrar mayores que 10, elevar al cuadrado y ordenar de mayor a menor
mayores_10_cuadrado_desc = sorted(
    map(lambda x: x ** 2, filter(lambda x: x > 10, numeros)),
    reverse=True
)

print("a.", pares_al_cuadrado)
print("b.", cubos_mayores_100_str)
print("c.", mayores_10_cuadrado_desc)
```
