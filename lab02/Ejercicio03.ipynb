```python
import math

# a. Convertir lista de temperaturas Celsius a Fahrenheit
celsius = [0, 15, 25, 100]
fahrenheit = list(map(lambda c: (c * 9 / 5) + 32, celsius))

# b. Sumar elemento a elemento dos listas de igual longitud
lista_1 = [1, 2, 3, 4]
lista_2 = [10, 20, 30, 40]
suma_listas = list(map(lambda a, b: a + b, lista_1, lista_2))

# c. Calcular el área de círculos dada una lista de radios
radios = [1, 2, 3, 5]
areas = list(map(lambda r: round(math.pi * (r ** 2), 2), radios))

# d. Generar lista de tuplas (nombre, valor + 10%) desde un diccionario
precios = {"pan": 1000, "leche": 2500, "café": 5000}
precios_con_aumento = list(
    map(lambda item: (item[0], round(item[1] * 1.10, 2)), precios.items())
)

# e. Multiplicar todos los elementos de una matriz por 10
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
matriz_por_10 = list(
    map(lambda fila: list(map(lambda x: x * 10, fila)), matriz)
)

print("a.", fahrenheit)
print("b.", suma_listas)
print("c.", areas)
print("d.", precios_con_aumento)
print("e.", matriz_por_10)
```
